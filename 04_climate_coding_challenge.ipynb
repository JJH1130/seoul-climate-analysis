### Step 2B: Clean up time series plots by resampling

You may notice that your plot looks a little “fuzzy”. This happens when Python is trying to plot a value for every date, but the resolution of the image is too low to actually do that. You can address this issue by **resampling** the data, or summarizing it over a time period of your choice. In this case, we will resample annually, giving us one data point per year.

```python
### Resample daily temperature data to annual mean temperature
ann_climate_df = (
    climate_df_units
    .resample('YS')
    .mean()
)

ann_climate_df
```

* **Missing Values (late 1930s to early 1940s):** There is a noticeable gap in the time series during the late 1930s and early 1940s. This is likely a data anomaly caused by sensor malfunction or missing records during that historical period (e.g., around World War II).
* **Outliers (around the mid-1930s and 1990s):** There are a sharp upward spike and a downward drop around 1935 and 1995, respectively. This could reflect either extreme physical climate events (such as unusually cold winters or volcanic cooling effects) or a data collection issue.

## Step 3: Make an interactive plot
You can use the `.hvplot()` method with similar arguments to create an interactive plot:

1. Copy your plotting code into the cell below  
2. Replace `.plot` in your code with `.hvplot`
3. Check that your code follows PEP-8 guidelines

Now, you should be able to hover over data points and see their values!

```python
### plot the annual data interactively

hv.extension('bokeh')

hvplot_ann_climate=ann_climate_df.hvplot(
    y='temp_c',
    title='Annual average temperature in Boulder, Colorado',
    xlabel='Year',
    ylabel='Temperature (°C)',
    legend=False,
    color='teal'
)

hvplot_ann_climate
```

### Explore the data
* **Minimum mean annual temperature:** 3.731 °C in 1994
* **Maximum mean annual temperature:** 20.778 °C in 1937
